```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

Creating an Agent


```python
from langchain.agents import AgentExecutor, create_openai_tools_agent
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")
tools = []

# Use MessagesPlaceholder for agent_scratchpad
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant."),
        ("human", "{input}"),
        MessagesPlaceholder(variable_name="agent_scratchpad"),
    ]
)

agent = create_openai_tools_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
agent
```

### GOOGLE Gemini Integration

```python
import os 
from langchain_google_genai import ChatGoogleGenerativeAI

# Initialize the model (using the correct provider prefix and valid model name)
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite", # Or "gemini-3.5-flash-lite" depending on your API access
    model_provider="google_genai"
)

# Invoke the model
response = model.invoke("Why do parrots talk?")

# Print the content
print(response.content)
```

### Groq Integration

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load keys from .env
load_dotenv()

# Initialize Groq model (provider:model syntax or explicit model_provider)
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Invoke
response = model.invoke("What is AGI?")
print(response.content)
```
