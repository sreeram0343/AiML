### Tools

In LangChain, a Tool is a function/API that an LLM or agent can call to perform an action or retrieve information that the model itself cannot directly access.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:llama-3.3-70b-versatile")
response = model.invoke("Why do parrots talk?")

print(response.content)
```

```python
from langchain.tools import tool

@tool 
def get_weather(location:str)->str:          #doc string """ """, a string written inside a function, class, or module to describe what it does
    """Get the weather at the location"""  
    return f"It's sunny in {location}"


model_with_tools = model.bind_tools([get_weather])
```

```python
tool_result = get_weather.invoke(
    response.tool_calls[0]["args"]
)

print(tool_result)
```

A Tool is a function, API, or capability that allows an LLM to perform a specific task or retrieve information. LangChain connects the LLM to those tools and manages the tool-calling workflow.
